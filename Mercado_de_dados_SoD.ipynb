## Insights primeira visualização



**Volume:** o dataset tem 3.495 respondentes e 388 colunas. Esse número de
colunas é indício de que a pesquisa usa perguntas de múltipla escolha, e cada
alternativa foi expandida em uma coluna própria (dummy). Cada bloco numerado (1 a 8) representa uma seção temática do questionário (perfil, empresa/vaga, gestão, tecnologia, etc.), e dentro de cada seção, as perguntas de múltipla escolha se abrem em colunas filhas.

**Observação 1 — volume de colunas:** olhando os nomes das colunas (ex:
`1.e.1_...`, `1.e.2_...`, `1.e.3_...`), percebe-se um padrão de numeração
hierárquica (pergunta.subpergunta.alternativa). Isso confirma a hipótese:
uma única pergunta de múltipla escolha (ex: "1.e") se desdobra em várias
colunas binárias, uma para cada alternativa possível.

**Observação 2 — padrão de NaN:** em várias linhas (ex: respondentes 0, 1, 2
e 9), todas as colunas a partir de um certo ponto do questionário aparecem
como `NaN` simultaneamente, em bloco — não de forma espalhada/aleatória.
Isso sugere que o `NaN` aqui não representa "não sei" ou dado ausente por
erro de coleta, e sim que o respondente **não viu aquela pergunta**, porque
a pesquisa tem lógica de pulo de seção (skip pattern) — por exemplo, alguém
que respondeu não trabalhar com dados pode não ter sido exposto às perguntas
técnicas mais adiante.

**Implicação para a análise:** se essa hipótese se confirmar, o tratamento de
nulos não pode ser genérico (ex: `dropna()` ou preencher com 0 direto). Será
necessário distinguir "não respondeu porque pulou a seção" de "não respondeu
por outro motivo", já que isso afeta diretamente o denominador correto para
qualquer cálculo de percentual mais adiante na análise.

**Próximo passo:** validar essas duas hipóteses com uma inspeção estrutural
mais formal (dtypes, proporção de nulos por coluna, e checagem por grupos de
colunas relacionadas).

### Tipos de dados e padrão de nulos

O dataset é composto por 315 colunas `float64`, 71 `object` e 2 `int64`.
A predominância de float64 confirma que a maior parte das colunas (quase 80%)
são variáveis binárias (0/1/NaN) — resultado da decomposição de perguntas de
múltipla escolha em colunas dummy.

O ranking de nulos por coluna reforça essa leitura: os maiores percentuais de
`NaN` aparecem concentrados em seções específicas do questionário (ex: bloco 8
— rotina de Cientista de Dados, ~86%; bloco 5 — busca de oportunidade,
88-95%), e não distribuídos aleatoriamente pelo dataset. Também aparece um
caso de nulo condicional dentro de um bloco geral (`1.h_pais_onde_mora`,
96,5%), sugerindo dependência de uma resposta anterior.

Isso indica que o `NaN`, neste dataset, é majoritariamente estrutural
(skip pattern da pesquisa) — não dado ausente por erro de coleta. Essa
hipótese ainda precisa ser validada formalmente antes de definir a
estratégia de tratamento de nulos.

```python
# Testa se o NaN é consistente dentro de um grupo (ex: bloco 8.b)
grupo_8b = [c for c in df.columns if c.startswith('8.b.')]

# Conta, por linha, quantas colunas do grupo estão nulas
nulos_por_linha = df[grupo_8b].isna().sum(axis=1)

print(nulos_por_linha.value_counts())
```

```python
# Repete o teste de consistência no bloco 4.c (linguagens de programação)
grupo_4c = [c for c in df.columns if c.startswith('4.c.')]

nulos_por_linha_4c = df[grupo_4c].isna().sum(axis=1)

print(len(grupo_4c))
print(nulos_por_linha_4c.value_counts())
```

### Validação do padrão de nulos por skip pattern

Repetindo o teste de consistência em dois blocos distintos — 8.b (técnicas
de Data Science, n=490 respostas) e 4.c (linguagens de programação, n=2.096
respostas) — o resultado se repete: dentro de um mesmo grupo de colunas
filhas, cada respondente ou preencheu **todas** as colunas do grupo, ou
**nenhuma**. Não há casos intermediários.

Isso confirma que o `NaN`, nas colunas dummy de múltipla escolha, representa
"não foi exposto à pergunta" (skip pattern da pesquisa) — e não recusa ou
erro de preenchimento em uma alternativa específica.

**Implicação prática para o restante da análise:** ao calcular qualquer
percentual sobre colunas dummy (ex: "% que usa Python"), o denominador correto
não é o total de 3.495 respondentes, e sim o número de pessoas que efetivamente
respondeu aquela seção.
